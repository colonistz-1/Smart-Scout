## Retrieving Premier League Players

```python
response = requests.get(
    f"{BASE_URL}/football/teams",
    headers=headers,
    params={
        "competition_id": "comp_3039",
        "season_id": "sn_6125938",
        "per_page": 100
    },
    timeout=20
)

response.raise_for_status()

teams_df = pd.json_normalize(response.json()["data"])

team_ids = teams_df["id"].tolist()

print("Total teams:", len(team_ids))
display(teams_df[["id", "name"]])
```

```python
'''https://www.thestatsapi.com/docs/concepts/rate-limits'''
def get_with_retry(path, params=None, max_retries=3):
    for attempt in range(max_retries + 1):

        response = requests.get(
            f"{BASE_URL}{path}",
            headers=headers,
            params=params,
            timeout=20
        )

        data = response.json()

        if response.ok:
            return data

        error = data.get("error", {})
        code = error.get("code")

        if code == "USAGE_LIMIT_EXCEEDED":
            raise RuntimeError(
                "Monthly API quota exhausted. Stop requesting."
            )

        if code == "RATE_LIMITED" and attempt < max_retries:
            wait = max(1, int(response.headers.get("Retry-After", 10)))
            print(f"Rate limited. Waiting {wait} seconds...")
            time.sleep(wait + 1)
            continue

        raise RuntimeError(
            f"{response.status_code}: {code} - {error.get('message')}"
        )
```

```python
all_midfielders = []

for i, team_id in enumerate(team_ids):

    data = get_with_retry(
        "/football/players",
        params={
            "team_id": team_id,
            "position": "M",
            "per_page": 100
        }
    )

    all_midfielders.extend(data["data"])

    print(f"{i+1}/{len(team_ids)} teams completed")

midfielders_df = pd.json_normalize(all_midfielders)

midfielders_df = midfielders_df.drop_duplicates(
    subset="id"
)

print("Total midfielders:", len(midfielders_df))

display(midfielders_df.head())
```
