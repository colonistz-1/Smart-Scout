```python
import sys
sys.path.insert(0, "..")  # make src/ importable from notebooks/

import matplotlib.pyplot as plt
import seaborn as sns

from src.data import load_data

df = load_data("../data/iris.csv")  # falls back to built-in Iris if the CSV is absent
df.head()
```

## Basic checks

Shape, dtypes, missing values, and summary statistics — always the first thing to look at.

```python
print(f"Shape: {df.shape}")
print(f"\nMissing values per column:\n{df.isna().sum()}")
df.describe()
```

## Class balance

Imbalanced targets change which metrics you should trust (accuracy becomes misleading).

```python
df["species"].value_counts().plot(kind="bar", title="Class balance")
plt.tight_layout()
plt.show()
```

## Feature relationships

A pairplot gives a quick view of which features separate the classes.

```python
sns.pairplot(df, hue="species", corner=True)
plt.show()
```
