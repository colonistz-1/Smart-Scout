## Retrieving Premier League Players

```python
response = requests.get(
    f"{BASE_URL}/football/teams",
    headers=headers,
    params={
        "competition_id": "comp_3039",
        "season_id": "sn_6125938",
        "per_page": 100
    },
    timeout=20
)

response.raise_for_status()

teams_df = pd.json_normalize(response.json()["data"])

team_ids = teams_df["id"].tolist()

print("Total teams:", len(team_ids))
display(teams_df[["id", "name"]])
```

```python
import time

'''https://www.thestatsapi.com/docs/concepts/rate-limits'''
def get_with_retry(path, params=None, max_retries=3):
    for attempt in range(max_retries + 1):

        response = requests.get(
            f"{BASE_URL}{path}",
            headers=headers,
            params=params,
            timeout=20
        )

        data = response.json()

        if response.ok:
            return data

        error = data.get("error", {})
        code = error.get("code")

        if code == "USAGE_LIMIT_EXCEEDED":
            raise RuntimeError(
                "Monthly API quota exhausted. Stop requesting."
            )

        if code == "RATE_LIMITED" and attempt < max_retries:
            wait = max(1, int(response.headers.get("Retry-After", 10)))
            print(f"Rate limited. Waiting {wait} seconds...")
            time.sleep(wait + 1)
            continue

        raise RuntimeError(
            f"{response.status_code}: {code} - {error.get('message')}"
        )
```

```python
all_midfielders = []

for i, team_id in enumerate(team_ids):

    data = get_with_retry(
        "/football/players",
        params={
            "team_id": team_id,
            "position": "M",
            "per_page": 100
        }
    )

    all_midfielders.extend(data["data"])

    print(f"{i+1}/{len(team_ids)} teams completed")

midfielders_df = pd.json_normalize(all_midfielders)

midfielders_df = midfielders_df.drop_duplicates(
    subset="id"
)

print("Total midfielders:", len(midfielders_df))

display(midfielders_df.head())
```

```python
Path("../data/raw").mkdir(parents=True, exist_ok=True)

# Save the dataset
midfielders_df.to_csv(
    "../data/raw/premier_league_midfielders.csv",
    index=False
)

print("Dataset saved successfully!")
```

```python
import json

output = Path("../data/raw/midfielder_stats_2025_26.jsonl")
output.parent.mkdir(parents=True, exist_ok=True)

# Read previously saved players
saved_ids = set()

if output.exists():
    with output.open("r", encoding="utf-8") as f:
        for line in f:
            saved_ids.add(json.loads(line)["player_id"])

for i, player in midfielders_df.iterrows():

    player_id = player["id"]
    player_name = player["name"]

    if player_id in saved_ids:
        continue

    try:
        result = get_with_retry(
            f"/football/players/{player_id}/stats",
            params={
                "season_id": "sn_6125938",
                "competition_id": "comp_3039"
            }
        )

        stats = result["data"]

        if stats is not None:
            stats["player_id"] = player_id
            stats["player_name"] = player_name

            with output.open("a", encoding="utf-8") as f:
                f.write(json.dumps(stats) + "\n")

            saved_ids.add(player_id)

        print(f"{i+1}/{len(midfielders_df)}: {player_name} ✓")

    except RuntimeError as e:
        if "404" in str(e):
            print(f"No 2025/26 stats: {player_name}")
            continue

        print(f"Stopped at {player_name}: {e}")
        break
```

```python
stats_df = pd.read_json(
    "../data/raw/midfielder_stats_2025_26.jsonl",
    lines=True
)

stats_df = pd.json_normalize(
    stats_df.to_dict(orient="records")
)

columns = [
    "player_name",
    "minutes_played",
    "scoring.goals",
    "scoring.assists",
    "scoring.expected_goals",
    "scoring.expected_assists",
    "passing.key_passes",
    "passing.pass_accuracy",
    "defending.tackles",
    "defending.interceptions"
]

display(stats_df[columns].head(10))
```

```python
stats_df.to_csv(
    "../data/raw/premier_league_stats_2025_26.csv",
    index=False
)

print("CSV saved!")
```
